## Parte 4 — Estadísticos

```python
print("Media:   ", df["Ventas"].mean())
print("Mediana: ", df["Ventas"].median())
print("Mínimo:  ", df["Ventas"].min())
print("Máximo:  ", df["Ventas"].max())
print("Suma:    ", df["Ventas"].sum())
print("Conteo:  ", df["Ventas"].count())
```

```python
# Apoyo para las reflexiones: cuánto aporta cada producto y cada región al total
por_producto = df.groupby("Producto")["Ventas"].agg(["count", "sum"])
por_producto["% del total"] = (por_producto["sum"] / df["Ventas"].sum() * 100).round(1)
print(por_producto.sort_values("sum", ascending=False))
print((df.groupby("Region")["Ventas"].sum() / df["Ventas"].sum() * 100).round(1))
```
